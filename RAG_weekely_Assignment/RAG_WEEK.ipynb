```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
model = SentenceTransformer('all-MiniLM-L6-v2')
sentence =  [
    "V is a mercenary in Night City who becomes connected to Johnny Silverhand after a biochip containing Johnny's engram is implanted in V's head.",

    "Johnny Silverhand is a rebellious rockerboy and former member of Samurai who strongly opposes the corporations controlling Night City.",

    "As the Relic begins overwriting V's personality with Johnny's engram, V searches for a way to remove the biochip before losing their identity.",

    "V and Johnny initially distrust each other, but their shared struggle gradually forces them to work together against Arasaka.",

    "Johnny's memories reveal his history with Arasaka, his relationship with Alt Cunningham, and his role in the destruction of Arasaka Tower."
]
chunkembedding = model.encode(sentence)

def take_que(query):
    embedding = model.encode([query])
    similarities = cosine_similarity(embedding,chunkembedding )[0]
    top_indices = similarities.argsort()[-3:][::-1]
    print("Query:", query)
    print("\nTop 3 results:")

    for index in top_indices:
        print(f"{similarities[index]:.4f} - {sentence[index]}")



take_que("Why is Johnny inside V's head?")
take_que("What is happening to V's identity?")

```
