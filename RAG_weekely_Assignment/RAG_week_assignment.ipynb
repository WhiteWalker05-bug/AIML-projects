```python
def chunnking():
  chunk=("""Jon snow was the king in the north."""
  """jon snow was a targeareyan.""")
  chunks=[]
  for i in range(0,len(chunk),17):
     chunks.append(chunk[i:i+17])
  return chunks
chunnking()
```

for this code i chose 17 as a specific number of words i wanted in a chunk because the lenght of the sentence itself is good enough for the words to properly fit in the chunk and in order for the LLM to understand it more better. i chose only one word "Jon Snow" as overlap because the snetences are related to him so it would be easier for the LLM to understand the context
