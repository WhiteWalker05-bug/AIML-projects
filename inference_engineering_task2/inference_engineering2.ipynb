```python
x=[-2.0,0.0,1.0,3.0]

qmin=-128
qmax=127

xmin=min(x)
xmax=max(x)

scale=(xmax-xmin)/(qmax-qmin)

zero_point=round(qmin-xmin/scale)

quantized=[]

for value in x:
  q=round(value/scale+zero_point)
  quantized.append(q)


print('original value:', x )
print('quantized value:', quantized)

```

```python
def model(parameters,dtype):
  if dtype=='fp32':
    bytes_per_parameter=4
  elif dtype=='fp16':
    bytes_per_parameter=2
  elif dtype=='int8' :
    bytes_per_parameter=1



  size_bytes=parameters * bytes_per_parameter

  size_md=size_bytes/(1024*1024)
  size_gb=size_bytes/(1024*1024*1024)
  return size_md,size_gb

size = model(1000000000, "fp32")

print("Size in MB:", size[0])
print("Size in GB:", size[1])
```

```python
def check_bound(flops, compute, bytes_needed, bandwidth):

    compute_time = flops / compute
    memory_time = bytes_needed / bandwidth

    if compute_time > memory_time:
        return "Compute-bound"
    else:
        return "Memory-bound"



print(check_bound(
      flops=100_000_000_000,
      compute=10_000_000_000_000,
      bytes_needed=1_000_000_000,
      bandwidth=1_000_000_000_000
))

print(check_bound(
    flops=5_000_000_000_000,
    compute=10_000_000_000_000,
    bytes_needed=500_000_000,
     bandwidth=1_000_000_000_000
))


print(check_bound(
    flops=1_000_000_000_000,
    compute=10_000_000_000_000,
    bytes_needed=50_000_000_000,
    bandwidth=1_000_000_000_000
))
```
