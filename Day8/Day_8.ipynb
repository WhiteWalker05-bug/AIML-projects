```python
numbers=[12,7,9,4,22,15,8,3,30,11]
count_even=0
count_odd=0
for i in numbers:

  if i%2==0:
    print("this number is even: ", i)
    count_even +=1
  else:
    print("this number is odd: ", i)
    count_odd+=1
  even_list=[i for i in numbers if i%2==0]

print("the number of even numbers is: ", count_even)
print("the number of odd numbers is: ", count_odd)
print(even_list)
```

```python
def calculate_average(numbers:list)->float:
 try:

    return sum(numbers)/len(numbers)
 except ZeroDivisionError:
    print("cannot divide by zero")

print(calculate_average([10,20,30]))
print(calculate_average([]))
```

```python
class Student:
    def __init__(self, name: str, marks: list):
        self.name = name
        self.marks = marks

    def average(self) -> float:
        return sum(self.marks) / len(self.marks)
student=  Student("adi", [80, 90, 85])
print(student.name)
print(student.average())
```
